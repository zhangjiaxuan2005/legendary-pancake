## 自注意力机制

```python
import torch

x1 = torch.tensor([1.1, 1.2, 1.3]).float()
x2 = torch.tensor([2.1, 2.2, 2.3]).float()
x3 = torch.tensor([3.1, 3.2, 3.3]).float()
x4 = torch.tensor([4.1, 4.2, 4.3]).float()

wq = torch.rand(3, 3)
wk = torch.rand(3, 3)
wv = torch.rand(3, 3)
```

```python
q1 = torch.matmul(x1, wq)
q2 = torch.matmul(x2, wq)
q3 = torch.matmul(x3, wq)
q4 = torch.matmul(x4, wq)

k1 = torch.matmul(x1, wk)
k2 = torch.matmul(x2, wk)
k3 = torch.matmul(x3, wk)
k4 = torch.matmul(x4, wk)

v1 = torch.matmul(x1, wv)
v2 = torch.matmul(x2, wv)
v3 = torch.matmul(x3, wv)
v4 = torch.matmul(x4, wv)
```

```python
self_attention_score11 = torch.matmul(q1, k1.T)
self_attention_score12 = torch.matmul(q1, k2.T)
self_attention_score13 = torch.matmul(q1, k3.T)
self_attention_score14 = torch.matmul(q1, k4.T)

self_attention_score21 = torch.matmul(q2, k1.T)
self_attention_score22 = torch.matmul(q2, k2.T)
self_attention_score23 = torch.matmul(q2, k3.T)
self_attention_score24 = torch.matmul(q2, k4.T)

self_attention_score31 = torch.matmul(q3, k1.T)
self_attention_score32 = torch.matmul(q3, k2.T)
self_attention_score33 = torch.matmul(q3, k3.T)
self_attention_score34 = torch.matmul(q3, k4.T)

self_attention_score41 = torch.matmul(q4, k1.T)
self_attention_score42 = torch.matmul(q4, k2.T)
self_attention_score43 = torch.matmul(q4, k3.T)
self_attention_score44 = torch.matmul(q4, k4.T)
```

```python
dk1 = k1.size(-1)
dk2 = k2.size(-1)
dk3 = k3.size(-1)
dk4 = k4.size(-1)

self_attention_score11 = self_attention_score11 / torch.sqrt(torch.tensor(dk1))
self_attention_score12 = self_attention_score12 / torch.sqrt(torch.tensor(dk1))
self_attention_score13 = self_attention_score13 / torch.sqrt(torch.tensor(dk1))
self_attention_score14 = self_attention_score14 / torch.sqrt(torch.tensor(dk1))

self_attention_score21 = self_attention_score21 / torch.sqrt(torch.tensor(dk2))
self_attention_score22 = self_attention_score22 / torch.sqrt(torch.tensor(dk2))
self_attention_score23 = self_attention_score23 / torch.sqrt(torch.tensor(dk2))
self_attention_score24 = self_attention_score24 / torch.sqrt(torch.tensor(dk2))

self_attention_score31 = self_attention_score31 / torch.sqrt(torch.tensor(dk3))
self_attention_score32 = self_attention_score32 / torch.sqrt(torch.tensor(dk3))
self_attention_score33 = self_attention_score33 / torch.sqrt(torch.tensor(dk3))
self_attention_score34 = self_attention_score34 / torch.sqrt(torch.tensor(dk3))

self_attention_score41 = self_attention_score41 / torch.sqrt(torch.tensor(dk4))
self_attention_score42 = self_attention_score42 / torch.sqrt(torch.tensor(dk4))
self_attention_score43 = self_attention_score43 / torch.sqrt(torch.tensor(dk4))
self_attention_score44 = self_attention_score44 / torch.sqrt(torch.tensor(dk4))
```

```python
self_attention_weight_q1 = torch.softmax(
    torch.tensor([self_attention_score11, self_attention_score12, self_attention_score13, self_attention_score14]),
    dim=0)
self_attention_weight_q2 = torch.softmax(
    torch.tensor([self_attention_score21, self_attention_score22, self_attention_score23, self_attention_score24]),
    dim=0)
self_attention_weight_q3 = torch.softmax(
    torch.tensor([self_attention_score31, self_attention_score32, self_attention_score33, self_attention_score34]),
    dim=0)
self_attention_weight_q4 = torch.softmax(
    torch.tensor([self_attention_score41, self_attention_score42, self_attention_score43, self_attention_score44]),
    dim=0)

```

### pytorch transformer

```python
import torch.nn as nn
import torch


class SelfAttention(nn.Module):
    def __init__(self, embed_dim: int, attn_dim: int, output_dim: int):
        super().__init__()
        self.embed_dim = embed_dim
        self.attn_dim = attn_dim
        self.output_dim = output_dim

        self.q_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)
        self.k_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)
        self.v_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)

        self.out_proj = nn.Linear(self.attn_dim, self.output_dim, bias=False)

    def forward(self, x):
        batch_size, seq_len, embed_dim = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        attn_score = torch.matmul(q, k.transpose(-2, -1))

        q_k = k.size(-1)
        attn_score = attn_score / torch.sqrt(torch.tensor(q_k))

        attn_weight = torch.softmax(attn_score, dim=-1)
        o = torch.matmul(attn_weight, v)

        return self.out_proj(o)
```

```python
x = torch.rand(4, 3, 3)
model = SelfAttention(3, 3, 3)
o = model(x)
o
```

### 多头注意力机制

```python
import torch.nn as nn
import torch


class MultiHeadAttention(nn.Module):
    def __init__(self, embed_dim: int, attn_dim: int, output_dim: int, num_heads: int):
        super().__init__()
        self.embed_dim = embed_dim
        self.attn_dim = attn_dim
        self.output_dim = output_dim
        self.num_heads = num_heads
        self.head_dim = attn_dim // num_heads


        self.q_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)
        self.k_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)
        self.v_proj = nn.Linear(embed_dim, self.attn_dim, bias=False)

        self.out_proj = nn.Linear(self.attn_dim, self.output_dim, bias=False)

    def forward(self, x):
        batch_size, seq_len, embed_dim = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)
        k = k.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)
        v = v.view(batch_size, seq_len, self.num_heads, self.head_dim).transpose(1, 2)

        attn_score = torch.matmul(q, k.transpose(-2, -1))
        q_k = k.size(-1)
        attn_score = attn_score / torch.sqrt(torch.tensor(q_k))
        attn_weight = torch.softmax(attn_score, dim=-1)

        o = torch.matmul(attn_weight, v)
        return self.out_proj(o.transpose(1, 2).reshape(batch_size, seq_len, self.attn_dim))
```

```python
x = torch.rand(1, 4, 2)
model = MultiHeadAttention(2, 6, 8, 2)
o = model(x)
o
```
