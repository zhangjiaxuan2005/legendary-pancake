### 位置编码

```python
import torch

x=torch.rand(4,8)
x
```

```python
seq_len=4
d_model=8
```

```python
pe=torch.zeros(seq_len,d_model)
pe
```

```python
pos=torch.arange(0,seq_len)
pos
```

```python
pe[:,0]=torch.sin(pos/10000**(2*0/d_model))
pe[:,2]=torch.sin(pos/10000**(2*1/d_model))
pe[:,4]=torch.sin(pos/10000**(2*2/d_model))
pe[:,6]=torch.sin(pos/10000**(2*3/d_model))
pe
```

```python
pe[:,1]=torch.cos(pos/10000**(2*0/d_model))
pe[:,3]=torch.cos(pos/10000**(2*1/d_model))
pe[:,5]=torch.cos(pos/10000**(2*2/d_model))
pe[:,7]=torch.cos(pos/10000**(2*3/d_model))
pe
```

```python
import math
import torch

seq_len=4
d_model=8
pe=torch.zeros(seq_len,d_model)
position=torch.arange(0,seq_len,dtype=torch.float).unsqueeze(1)
div_term=torch.exp(torch.arange(0,d_model,2).float()*(-math.log(10000.0)/d_model))
pe[:,0::2]=torch.sin(position*div_term)
pe[:,1::2]=torch.cos(position*div_term)
pe
```
